#Relations and Properties

```python
def check_relation_properties(A,R):
  R_set = set(R)

  is_reflexive = all((x,x) in R_set for x in A)

  is_symmetric = all((y,x) in R_set for (x,y) in R_set)

  is_transitive = True
  for x,y in R_set:
    for y2,z in R_set:
      if y == y2 and (x,z) not in R_set:
        is_transitive = False
        break
    if not is_transitive:
      break

  return {
      "reflexive": is_reflexive,
      "symmetric": is_symmetric,
      "transitive": is_transitive
  }
```

```python
A = {1,2,3}
R = {(1,1),(1,3),(2,2),(3,3),(3,1)}
check_relation_properties(A,R)
```

```python
A = {"Ravi","Hari","Shyam"}
R = {("Ravi","Ravi")}
check_relation_properties(A,R)
```
