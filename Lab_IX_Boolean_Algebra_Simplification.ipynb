```python
from sympy import symbols, simplify_logic
from sympy.logic import SOPform, POSform
from sympy.abc import A, B, C
from schemdraw.parsing import logicparse
```

```python
A, B, C, D, E = symbols('A B C D E')

# Define minterms (rows where the output is True)
minterms = [0,1,8,9,16,17,20,21,24,25]


# Generate minimized SOP expression
minimized_sop = SOPform([A,B,C,D,E], minterms)
print("Minimized SOP from Truth Table:", minimized_sop)
```

```python
minimized_sop = str(minimized_sop)
```

```python
expr_str = minimized_sop
# logicparse handles placement, wires, and gate linkages automatically
with logicparse(expr_str, outlabel="$Y$") as d:
    d.save("optimized_circuit.png")
```
