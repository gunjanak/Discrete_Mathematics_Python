# Logic Operators in Python




```python
def NOT(p):
    return not p

# Example: Is the security alarm active?
is_door_open = True
trigger_alarm = NOT(is_door_open)

print(f"Is door open? {is_door_open}")
print(f"Should we trigger the alarm? {trigger_alarm}")
```

```python
def AND(p, q):
    return p and q

# Example: Accessing a secure lab room
has_id_card = True
knows_password = False

access_granted = AND(has_id_card, knows_password)
print(f"Access granted? {access_granted}")
```

```python
def OR(p, q):
    return p or q

# Example: Unlocking a smartphone
fingerprint_matched = False
face_id_matched = True

phone_unlocks = OR(fingerprint_matched, face_id_matched)
print(f"Does the phone unlock? {phone_unlocks}")
# Output: True (because face_id_matched is True)
```

```python
def IMPLIES(p, q):
    return not p or q

# Example: A professor's rule: "If you submit late (P), you lose points (Q)"
submitted_late = True
lost_points = False

# Did the rule hold true?
rule_valid = IMPLIES(submitted_late, lost_points)
print(f"Did the grading rule hold up? {rule_valid}")
# Output: False (The student was late, but didn't lose points, breaking the rule)
```

```python
def BICONDITIONAL(p, q):
    return p == q

# Example: Equal rights / balance check
left_sensor_high = False
right_sensor_high = False

system_balanced = BICONDITIONAL(left_sensor_high, right_sensor_high)
print(f"Is the system balanced? {system_balanced}")
# Output: True (Both are False, meaning they match perfectly)
```

#Truth table generation

```python
# Define our simple logic functions
def NOT(p):
    return not p

def AND(p, q):
    return p and q

# 1. Print the header of our table
print("P     | Q     | P AND (NOT Q)")
print("-" * 30)

# 2. Use nested loops to generate all 4 combinations of True/False
for p in [True, False]:
    for q in [True, False]:

        # Calculate the result of our expression
        result = AND(p, NOT(q))

        # 3. Print the row (formatted with spaces for alignment)
        print(f"{str(p):<5} | {str(q):<5} | {str(result)}")
```

```python
# 1. Print the header of our table
print("P     | Q     | R     | (P AND Q) OR R")
print("-" * 38)

# 2. Triple nested loops to generate all 8 combinations
for p in [True, False]:
    for q in [True, False]:
        for r in [True, False]:

            # Calculate (P AND Q)
            p_and_q = AND(p, q)

            # Calculate the final result: (P AND Q) OR R
            result = OR(p_and_q, r)

            # 3. Print the row neatly aligned
            print(f"{str(p):<5} | {str(q):<5} | {str(r):<5} | {str(result)}")
```

```python
# Simple helper functions
def NOT(p): return not p
def AND(p, q): return p and q
def OR(p, q): return p or q

# 1. Print the header
print("P     | Q     | R     | P OR (Q AND (NOT R))")
print("-" * 44)

# 2. Triple nested loops for 8 combinations
for p in [True, False]:
    for q in [True, False]:
        for r in [True, False]:

            # Calculate (NOT R)
            not_r = NOT(r)

            # Calculate Q AND (NOT R)
            q_and_not_r = AND(q, not_r)

            # Calculate the final result
            result = OR(p, q_and_not_r)

            # 3. Print the row neatly
            print(f"{str(p):<5} | {str(q):<5} | {str(r):<5} | {str(result)}")
```

```python
# Simple helper functions
def AND(p, q): return p and q
def IMPLIES(p, q): return not p or q

# 1. Print the header
print("P     | Q     | R     | (P AND Q) IMPLIES R")
print("-" * 44)

# 2. Triple nested loops for 8 combinations
for p in [True, False]:
    for q in [True, False]:
        for r in [True, False]:

            # Calculate (P AND Q)
            p_and_q = AND(p, q)

            # Calculate final implication
            result = IMPLIES(p_and_q, r)

            # 3. Print the row neatly
            print(f"{str(p):<5} | {str(q):<5} | {str(r):<5} | {str(result)}")
```

#Tautology and Contradiction Checker

```python
# Simple helper functions
def NOT(p): return not p
def OR(p, q): return p or q
def AND(p, q): return p and q

# Let's test: P OR (NOT P)
print("Testing: P OR (NOT P)")

results = []
for p in [True, False]:
    # Evaluate the expression
    outcome = OR(p, NOT(p))
    results.append(outcome)
    print(f"P = {str(p):<5} | Result = {outcome}")

# Check the final status
if all(results):
    print("--> Classification: TAUTOLOGY (Always True!)\n")
elif not any(results):
    print("--> Classification: CONTRADICTION (Always False!)\n")
else:
    print("--> Classification: CONTINGENCY (Mixed)\n")
```

#De Morgan's Law Verification

```python
def NOT(p): return not p
def AND(p, q): return p and q
def OR(p, q): return p or q

print("Verifying De Morgan's Law: NOT (P AND Q) == (NOT P) OR (NOT Q)")
print("-" * 55)
print("P     | Q     | LHS: NOT (P AND Q) | RHS: (NOT P) OR (NOT Q) | Match?")
print("-" * 55)

all_match = True

for p in [True, False]:
    for q in [True, False]:
        # Left-Hand Side
        lhs = NOT(AND(p, q))

        # Right-Hand Side
        rhs = OR(NOT(p), NOT(q))

        # Do they match?
        match = (lhs == rhs)
        if not match:
            all_match = False

        print(f"{str(p):<5} | {str(q):<5} | {str(lhs):<18} | {str(rhs):<22} | {match}")

print("-" * 55)
if all_match:
    print("SUCCESS: De Morgan's Law is fully verified!")
else:
    print("FAIL: The sides do not match.")
```
