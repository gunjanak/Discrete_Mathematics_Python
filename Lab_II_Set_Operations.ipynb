#Basic Set Operations (Union, Intersection, Difference)

```python
# Define two sample sets
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# 1. Union (Elements in A OR B)
union_set = A.union(B)  # Or: A | B

# 2. Intersection (Elements in both A AND B)
intersection_set = A.intersection(B)  # Or: A & B

# 3. Difference (Elements in A but NOT in B)
difference_A_B = A.difference(B)  # Or: A - B

print(f"Set A: {A}")
print(f"Set B: {B}\n")
print(f"Union (A ∪ B): {union_set}")
print(f"Intersection (A ∩ B): {intersection_set}")
print(f"Difference (A - B): {difference_A_B}")
```

#Symmetric Difference and Subset Checks

```python
A = {1, 2, 3}
B = {1, 2, 3, 4, 5}

# 1. Symmetric Difference (Elements in A or B, but NOT both)
sym_diff = A.symmetric_difference(B)  # Or: A ^ B

# 2. Subset Check (Is A inside B?)
is_subset = A.issubset(B)

# 3. Superset Check (Does B contain A?)
is_superset = B.issuperset(A)

print(f"Symmetric Difference (A ▵ B): {sym_diff}")
print(f"Is A a subset of B? {is_subset}")
print(f"Is B a superset of A? {is_superset}")
```

#Cartesian Product of Two Sets

```python
A = {1, 2}
B = {'red', 'blue'}

cartesian_product = []

# Use nested loops to pair every element
for a in A:
    for b in B:
        cartesian_product.append((a, b))

print(f"Set A: {A}")
print(f"Set B: {B}")
print(f"Cartesian Product (A × B): {cartesian_product}")
```

#Power Set Generator

```python
def power_set(s):
    s = list(s)
    n = len(s)
    result = []
    for i in range(2**n):
        subset = [s[j] for j in range(n) if (i >> j) & 1]
        result.append(subset)

    result_set = set(tuple(x) for x in result)
    # print(set(tuple(x) for x in result))  # tuples are hashable
    return result_set

a = {1, 2, 3}
print(power_set(a))
```

#Venn Diagram Visualization

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# Define our sets
A = {"j","h","a",'r','e',"n"}
B = {"a","e","i","o","u"}

# Create the Venn diagram
# set_labels gives titles to the circles
# subset_sizes handles the overlaps automatically based on your sets
plt.figure(figsize=(6, 4))
venn2([A, B], set_labels=('Set A', 'Set B'))

# Display the diagram window
plt.title("Venn Diagram of Sets A and B")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn3

A = {1, 2, 3, 4}
B = {3, 4, 5, 6, 7}
C = {4, 7, 8, 9}

plt.figure(figsize=(6, 6))
venn3([A, B, C], set_labels=('Set A', 'Set B', 'Set C'))
plt.title("Venn Diagram of Sets A, B, and C")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

A = {1, 2, 3, 4}
B = {3, 4, 5, 6, 7}

plt.figure(figsize=(6, 4))
v = venn2([A, B], set_labels=('Set A', 'Set B'))

# Customize colors
v.get_patch_by_id('10').set_color('skyblue')   # only in A
v.get_patch_by_id('01').set_color('lightgreen') # only in B
v.get_patch_by_id('11').set_color('yellow')     # intersection

plt.title("Custom-Colored Venn Diagram")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2_circles

A = {1, 2, 3, 4}
B = {3, 4, 5, 6, 7}

plt.figure(figsize=(6, 4))
venn2_circles([A, B], linestyle='dashed', linewidth=2, color='gray')
plt.title("Outline-Only Venn Diagram")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2

# (Ab, aB, AB) = (only A, only B, both)
plt.figure(figsize=(6, 4))
venn2(subsets=(4, 5, 2), set_labels=('Set A', 'Set B'))
plt.title("Venn Diagram from Subset Sizes")
plt.show()
```
