# Print Tree

```python
def print_tree(graph, node, prefix=""):
    print(prefix + node)
    children = graph.get(node, [])

    for i, child in enumerate(children):
        # Determine if this child is the last one in the list
        is_last = (i == len(children) - 1)
        connector = "└── " if is_last else "├── "
        extension = "    " if is_last else "│   "

        # Recursive call for children
        print_tree(graph, child, prefix + connector)
```

# Pre order traversal

```python
def preorder_traversal(graph, node, visited=None, result=None):
    if visited is None:
        visited = set()  # Track visited nodes to avoid cycles
    if result is None:
        result = []      # Store the traversal order

    if node not in visited:
        result.append(node)  # 1. Visit the root node first
        visited.add(node)

        # 2. Recurse for all children in order (e.g., left to right)
        for child in graph.get(node, []):
            preorder_traversal(graph, child, visited, result)

    return result
```

# Inorder Traversal

```python
def inorder_traversal(graph, node, visited=None, result=None):
    if visited is None:
        visited = set()
    if result is None:
        result = []

    if node not in visited:
        visited.add(node)
        children = graph.get(node, [])

        if not children:
            # If it's a leaf node, just append it
            result.append(node)
        else:
            # Split children roughly in half to mimic Left -> Root -> Right
            mid = len(children) // 2

            # 1. Traverse first half of children (like "Left")
            for i in range(mid):
                inorder_traversal(graph, children[i], visited, result)

            # 2. Visit Root Node
            result.append(node)

            # 3. Traverse second half of children (like "Right")
            for i in range(mid, len(children)):
                inorder_traversal(graph, children[i], visited, result)

    return result


```

# Postorder Traversal

```python
def postorder_traversal(graph, node, visited=None, result=None):
    if visited is None:
        visited = set()
    if result is None:
        result = []

    if node not in visited:
        visited.add(node)

        # 1. Traverse all children first (Left to Right)
        for child in graph.get(node, []):
            postorder_traversal(graph, child, visited, result)

        # 2. Visit Root Node last
        result.append(node)

    return result
```
